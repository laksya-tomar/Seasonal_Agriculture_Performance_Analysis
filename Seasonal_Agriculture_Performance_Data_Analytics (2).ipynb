#Fix Null Values <h6>-by median

```python

print("Nulls before:\n", dataset.isnull().sum()[dataset.isnull().sum() > 0])

# Fill all numeric columns' nulls with their median
numeric_cols = dataset.select_dtypes(include='number').columns
dataset[numeric_cols] = dataset[numeric_cols].fillna(dataset[numeric_cols].median())

# Check nulls after
print("\nNulls after:\n", dataset.isnull().sum().sum())

# Save the cleaned file
dataset.to_csv('seasonal_agriculture_performance_dataset_cleaned.csv', index=False)
```

```python
numeric_cols=dataset.select_dtypes(include=np.number).columns
print(numeric_cols)
```

```python
outlier_summary = []

for col in numeric_cols:
    Q1 = dataset[col].quantile(0.25)
    Q3 = dataset[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = dataset[(dataset[col] < lower_bound) | (dataset[col] > upper_bound)]

    outlier_summary.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers),
        'Outlier %': round((len(outliers) / len(dataset)) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    by='Outlier Count',
    ascending=False
)
```

#Correlation Analysis

```python
analysis_cols = [
    'Yield_Tonnes_Ha',
    'Profit_INR',
    'Water_Efficiency_t_per_1000m3',
    'Avg_Temperature_C',
    'Rainfall_mm',
    'Humidity_pct',
    'Water_Used_m3',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha'
]

correlation = dataset[analysis_cols].corr()

correlation.round(2)

plt.figure(figsize=(12,8))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Between Agricultural Variables')
plt.show()
```

#Three Student- designed analyses

```python
# ===== Student-Designed Analysis 1: Irrigation Method vs Yield =====
irrigation_yield = dataset.groupby('Irrigation_Method')['Yield_Tonnes_Ha'].mean().round(2).sort_values(ascending=False)
print(irrigation_yield)

plt.figure(figsize=(8,5))
sns.barplot(x=irrigation_yield.index, y=irrigation_yield.values, palette='mako')
plt.title('Average Yield by Irrigation Method')
plt.xlabel('Irrigation Method')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# ===== Student-Designed Analysis 2: Fertilizer Efficiency by Crop =====
dataset['Fertilizer_Efficiency'] = dataset['Yield_Tonnes_Ha'] / dataset['Fertilizer_kg_ha']

fert_efficiency = dataset.groupby('Crop')['Fertilizer_Efficiency'].mean().round(3).sort_values(ascending=False)
print(fert_efficiency)

plt.figure(figsize=(9,5))
sns.barplot(x=fert_efficiency.index, y=fert_efficiency.values, palette='crest')
plt.title('Yield per kg of Fertilizer Used, by Crop')
plt.xlabel('Crop')
plt.ylabel('Yield / Fertilizer (Tonnes per kg/ha)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ===== Student-Designed Analysis 3: Soil pH & Moisture Sweet Spot for Yield =====
plt.figure(figsize=(9,6))
sc = plt.scatter(
    dataset['Soil_pH'],
    dataset['Soil_Moisture_pct'],
    c=dataset['Yield_Tonnes_Ha'],
    cmap='YlGn',
    alpha=0.7,
    s=40
)
plt.colorbar(sc, label='Yield (Tonnes/Ha)')
plt.title('Soil pH vs Soil Moisture, Colored by Yield')
plt.xlabel('Soil pH')
plt.ylabel('Soil Moisture')
plt.tight_layout()
plt.show()

# Optional: bin soil pH into ranges to find the sweet spot
dataset['pH_Range'] = pd.cut(dataset['Soil_pH'], bins=5)
ph_yield = dataset.groupby('pH_Range')['Yield_Tonnes_Ha'].mean().round(2)
print(ph_yield)
```
